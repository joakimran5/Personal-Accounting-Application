```python
import pandas as pd
import numpy as np
import statistics
```

```python
# Convert data to matrix
df = pd.read_csv("transactions.csv")
df.head()
```

```python
df.groupby("tsc_descrp")["tsc_amt"].apply(list)
```

```python
petrol_array = df.loc[df["tsc_descrp"] == "petrol", "tsc_amt"].to_numpy()
petrol_array
```

# Measure of central tendency

```python
mean_value = np.round(np.mean(petrol_array), 2)
median_value = statistics.median(petrol_array)
mode_value = statistics.mode(petrol_array)

print("Mean gas consumption:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```
