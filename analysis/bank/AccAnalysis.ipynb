```python
import pandas as pd
import numpy as np
from scipy import linalg
```

```python
# Convert data to matrix
df = pd.read_csv("transactions.csv")
amounts = df["tsc_amt"].values
```

```python
# Matrix operations
X = pd.get_dummies(df["tsc_cat"]).values
y = amounts
```

```python
# Solve linear system
w = linalg.solve(X.T @ X, X.T @ y) # (XᵀX)w = Xᵀy
```

```python
# Matrix Decomposition
U, s, Vh = linalg.svd(X, full_matrices=False) # Singular Value Decomposition
```

```python
# Eigenvalues (pattern detection)
eigvals, eigvecs = linalg.eig(X.T @ X)
```

```python
U, s, Vh = linalg.svd(X)
X_reduced = U[:, :2] * s[:2]
```
